```python
arm = np.load("wrist_charge_times_ms.npy")
leg = np.load("leg_charge_times_ms.npy")
```

```python
plt.figure(figsize=(10, 3))

bins = np.linspace(0, 6.5, 15)
counts, edges = np.histogram(leg_model, bins=bins)
counts = counts / np.sum(counts)  # bar heights sum to 1
plt.bar(edges[:-1], counts, width=np.diff(edges), edgecolor='black', linewidth=2, alpha=0.35, label='Model Distribution')
counts, edges = np.histogram(leg/1000, bins=bins)
counts = counts / np.sum(counts)  # bar heights sum to 1
plt.bar(edges[:-1], counts, width=np.diff(edges), edgecolor='black', linewidth=2, alpha=0.35, label='Prototype Distribution')

plt.title("Distribution of Charge Times for Walking (Sensor on Leg)")
plt.xlabel("Charge Time [seconds]")
plt.ylabel("Probability")
plt.axvline((leg_model).mean(),c='tab:blue',linewidth=3,zorder=4,linestyle='--',label='Model Mean')
plt.axvline((leg/1000).mean(),c='tab:orange',linewidth=3,zorder=4,linestyle='--',label='Prototype Mean')
plt.grid(zorder=0)
plt.legend()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Example bins
bins = np.linspace(0, 6.5, 15)

# Define a common font size
TITLE_FONT = 20
LABEL_FONT = 20
TICK_FONT = 20
LEGEND_FONT = 20

# Create a figure with 2 rows, 1 column, shared x-axis
fig, axes = plt.subplots(1, 2, figsize=(18, 6), sharex=False,sharey=True)

# -----------------------------
# Top subplot: Leg
# -----------------------------
counts, edges = np.histogram(leg_model, bins=bins)
counts = counts / np.sum(counts)
axes[0].bar(edges[:-1], counts, width=np.diff(edges), edgecolor='black',
            linewidth=2, alpha=0.35, label=r'$\mathbf{Distribution}$: Energy Harvesting Model')

counts, edges = np.histogram(leg/1000, bins=bins)
counts = counts / np.sum(counts)
axes[0].bar(edges[:-1], counts, width=np.diff(edges), edgecolor='black',
            linewidth=2, alpha=0.35, label=r'$\mathbf{Distribution}$: Wearable Prototype')

axes[0].axvline(leg_model.mean(), c='tab:blue', linewidth=3,
                zorder=4, linestyle='--', label=r'$\mathbf{Mean}$: Energy Harvesting Model')
axes[0].axvline((leg/1000).mean(), c='tab:orange', linewidth=3,
                zorder=4, linestyle='--', label=r'$\mathbf{Mean}$: Wearable Prototype')

axes[0].set_title(r"Interarrival Distribution, $\mathbf{Walking}$ (Sensor on $\mathbf{Leg}$)",
                  fontsize=TITLE_FONT, fontweight='bold')
axes[0].set_xlabel("Interarrival Time [seconds]", fontsize=LABEL_FONT, fontweight='bold')
axes[0].set_ylabel("Probability", fontsize=LABEL_FONT, fontweight='bold')
axes[0].tick_params(axis='both', labelsize=TICK_FONT)
axes[0].grid(zorder=0)
fig.legend(fontsize=LEGEND_FONT,loc='center left', bbox_to_anchor=(0.3, 0.5))

# -----------------------------
# Bottom subplot: Arm
# -----------------------------
counts, edges = np.histogram(arm_model, bins=bins)
counts = counts / np.sum(counts)
axes[1].bar(edges[:-1], counts, width=np.diff(edges), edgecolor='black',
            linewidth=2, alpha=0.35, label=r'$\mathbf{Distribution}$: Energy Harvesting Model')

counts, edges = np.histogram((arm+500)/1000, bins=bins)
counts = counts / np.sum(counts)
axes[1].bar(edges[:-1], counts, width=np.diff(edges), edgecolor='black',
            linewidth=2, alpha=0.35, label=r'$\mathbf{Distribution}$: Wearable Prototype')

axes[1].axvline(arm_model.mean(), c='tab:blue', linewidth=3,
                zorder=4, linestyle='--', label=r'$\mathbf{Mean}$: Energy Harvesting Model')
axes[1].axvline(((arm+500)/1000).mean(), c='tab:orange', linewidth=3,
                zorder=4, linestyle='--', label=r'$\mathbf{Mean}$: Wearable Prototype')

axes[1].set_title(r"Interarrival Distribution, $\mathbf{Walking}$ (Sensor on $\mathbf{Arm}$)",
                  fontsize=TITLE_FONT, fontweight='bold')
axes[1].set_xlabel("Interarrival Time [seconds]", fontsize=LABEL_FONT, fontweight='bold')
# axes[1].set_ylabel("Probability", fontsize=LABEL_FONT, fontweight='bold')
axes[1].tick_params(axis='both', labelsize=TICK_FONT)
axes[1].grid(zorder=0)
# axes[1].legend(fontsize=LEGEND_FONT)

# Adjust spacing between subplots
plt.tight_layout()
plt.savefig('dist.svg')
# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
activity_path = 'GeffenHAR/Geffen_trial2/'
activity = np.load(activity_path+'lying-down.npy')

sensors = ['right_arm', 'left_arm', 'right_leg']

fig,ax = plt.subplots(1,3,figsize=(18,5),sharey=True)
for i in range(3):
    ax[i].plot(np.arange(activity.shape[0])/25,activity[:,i*3],label=f'X')
    ax[i].plot(np.arange(activity.shape[0])/25,activity[:,i*3+1],label=f'Y')
    ax[i].plot(np.arange(activity.shape[0])/25,activity[:,i*3+2],label=f'Z')
    ax[i].grid()
    ax[i].set_title(sensors[i])
    ax[i].legend()
```
